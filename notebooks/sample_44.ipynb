* 絶対位置埋め込みを利用（RoPEではない）

### 注意点
* 汎用的な内容は生成不能
* 学習した内容、学習した文と類似文章が入力されるとうまく生成される
* 学習していない内容は、全くだめ
* モデルサイズ・データサイズが小さいので完全コピーに近い生成がみられるがこれが正常な状況
* 系列長を256とある程度長くしないと、指示チューニング時にトークン不足となる
* 事前学習だけを考える場合、GPUメモリとの兼ね合いで seq_len=128でも事前学習による生成可能

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader # カスタムDataset作成
from tokenizers import Tokenizer
from pathlib import Path

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"{device=}")
```

```python
data_dir = "./data/homme_texts/"                             # 学習用データのディレクトリ
tokenizer_filename = "tokenizer/tokenizer_10k.json"   # tokenizer
pretrain_filename = "model/static_kv_homme_seq_256_bpe_10k.model"
tokenizer = Tokenizer.from_file(tokenizer_filename)         # トークナイザー
print(f"size: {tokenizer.get_vocab_size()}")
```

```python
class ModelConfig:
    def __init__(self, tokenizer):
        # モデル構造
        self.vocab_size = tokenizer.get_vocab_size()
        self.seq_len = 256   # 128トークンだとSFT時に少ない
        self.d_model = 256   # 512
        self.nhead = 8
        self.dim_feedforward = 4*self.d_model
        self.num_layers = 6
        self.dropout = 0.1

        # 特殊トークンID
        self.pad_token_id = tokenizer.token_to_id("<pad>")
        self.eod_token_id = tokenizer.token_to_id("<eod>")

        # 学習データに関する設定
        self.context_size = self.seq_len         # 学習できる長さ

        # 学習設定
        self.batch_size = 256
        self.learning_rate = 0.001  # これだとデフォルトと変わらない
        self.num_steps = 2500
        self.max_grad_norm = 1.0
        self.ignore_index = -100
        self.weight_decay = 0.01
```

```python
import math
import torch.nn.functional as F

class StaticKVCache:
    """
    Static KV Cache
    1. 事前に、全層の K/V を (batch, nhead, max_cache_len, head_dim) で確保
    2. Attention には常に max_cache_len の長さ、未使用部分はマスク
    3. 書き込みは cache_position の位置へ index_copy_ で 置き換え
    """
    def __init__(self, config, batch_size, device, dtype, max_cache_len=None):
        self.max_batch_size = batch_size
        self.max_cache_len = max_cache_len or config.seq_len
        head_dim = config.d_model // config.nhead
        shape = (batch_size, config.nhead, self.max_cache_len, head_dim)

        # (1) 0で予めKVキャッシュの領域をテンソルとして確保
        # forループ：すべてのTransformer層で作成 
        self.key_cache   = []
        self.value_cache = []

        for _ in range(config.num_layers):
            # empty ではなく zeros
            # 未書き込みの領域も Attention に入るぞ　計算量が減ることはない
            k = torch.zeros(shape, device=device, dtype=dtype)
            v = torch.zeros(shape, device=device, dtype=dtype)
            self.key_cache.append(k)
            self.value_cache.append(v)


    # (2) index_copy_() で、k、v cacheのテンソルの指定位置に値を上書き
    # layer_idxのキャッシュを戻り値とする
    def update(self, k, v, layer_idx, cache_position):
        """k, v: (bs, nhead, T, head_dim) を cache_position の位置へ書き込み"""
        k_out = self.key_cache[layer_idx]
        v_out = self.value_cache[layer_idx]
        k_out.index_copy_(2, cache_position, k)   # dim=2 (系列方向) に書き込み
        v_out.index_copy_(2, cache_position, v)
        return k_out, v_out

    # KとVのキャッシュを０で初期化（新しく作成ではなくて上書き）
    def reset(self):
        for k, v in zip(self.key_cache, self.value_cache):
            k.zero_()
            v.zero_()


class CausalSelfAttention(nn.Module):
    def __init__(self, config: ModelConfig, layer_idx: int):
        super().__init__()
        assert config.d_model % config.nhead == 0

        self.config = config
        self.layer_idx = layer_idx   # Transformerのlayer index
        self.nhead = config.nhead
        self.head_dim = config.d_model // config.nhead
        self.qkv = nn.Linear(config.d_model, 3 * config.d_model, bias=False)
        self.proj = nn.Linear(config.d_model, config.d_model, bias=False)
        self.attn_dropout_p = config.dropout
        self.resid_dropout = nn.Dropout(config.dropout)

    def forward(self, x, kv_cache=None, cache_position=None, attn_mask=None):
        bs, seq_len, d_model = x.shape
        # q, k, v を計算
        q, k, v = self.qkv(x).split(d_model, dim=2)
        q = q.view(bs, seq_len, self.nhead, self.head_dim).transpose(1, 2)
        k = k.view(bs, seq_len, self.nhead, self.head_dim).transpose(1, 2)
        v = v.view(bs, seq_len, self.nhead, self.head_dim).transpose(1, 2)

        # (4)
        if kv_cache is not None:
            # 推論時（prefill / decode）
            # キャッシュへ書き込み、形状 (bs, nhead, max_cache_len, head_dim) の K/V を受け取る
            k, v = kv_cache.update(k, v, self.layer_idx, cache_position)
            # 未来の位置・未書き込みの位置は attn_mask で隠す
            y = F.scaled_dot_product_attention(q, k, v, attn_mask=attn_mask)
        else:
            # 学習時: キャッシュなしの通常の causal attention
            y = F.scaled_dot_product_attention(
                q, k, v,
                is_causal=True,
                dropout_p=(self.attn_dropout_p if self.training else 0.0),
            )
        y = y.transpose(1, 2).contiguous().view(bs, seq_len, d_model)
        y = self.resid_dropout(self.proj(y))
        return y

# SwiGLUタイプのFFN
class FFN(nn.Module):
    def __init__(self, config):
        super().__init__()
        hidden = config.dim_feedforward
        self.w_gate = nn.Linear(config.d_model, hidden, bias=False)
        self.w_up   = nn.Linear(config.d_model, hidden, bias=False)
        self.w_down = nn.Linear(hidden, config.d_model, bias=False)
        self.activation = nn.SiLU()
        self.dropout = nn.Dropout(config.dropout)

    def forward(self, x):
        gate = self.activation(self.w_gate(x))    # ゲート側
        up   = self.w_up(x)                       # そのまま通過の特徴量
        return self.dropout(self.w_down(gate*up)) # 線形層でFFNと同様に形状を整える出力

# RMSNormタイプのTransformerブロック
class TransformerBlock(nn.Module):
    def __init__(self, config, layer_idx):
        super().__init__()

        self.attn_norm = nn.RMSNorm(config.d_model, eps=1e-6)
        self.attn = CausalSelfAttention(config, layer_idx)
        self.ffn_norm = nn.RMSNorm(config.d_model, eps=1e-6)
        self.ffn = FFN(config)

    def forward(self, x, kv_cache=None, cache_position=None, attn_mask=None):
        x = x + self.attn(self.attn_norm(x), kv_cache=kv_cache, cache_position=cache_position, attn_mask=attn_mask) # x + Attention(RMSNorm(x))
        x = x + self.ffn(self.ffn_norm(x))  # x + FFN(RMSNorm(x))
        return x
```

```python
class DNN(nn.Module):
    def __init__(self, config: ModelConfig):
        super().__init__()
        self.config = config

        # 絶対位置　埋め込み (RoPEは次回以降)
        self.token_embedding = nn.Embedding(num_embeddings=config.vocab_size, embedding_dim=config.d_model)
        self.pos_embedding = nn.Embedding(num_embeddings=config.seq_len, embedding_dim=config.d_model)
        self.dropout = nn.Dropout(config.dropout)

        # causal transformer layer
        # 各層のlayer_idxを指定、キャッシュ更新などで利用
        self.transformer_blocks = nn.ModuleList([TransformerBlock(config, i) for i in range(config.num_layers)])

        # 最後の出力に向けた正規化とFC層　最終的に単語数になる
        self.norm = nn.RMSNorm(config.d_model, eps=1e-6)
        self.fc = nn.Linear(config.d_model, config.vocab_size,bias=False)
        self.apply(self._init_weights)                    # 個別に初期化
        self.fc.weight = self.token_embedding.weight      # 重み共有


        # 残差方向の出力層だけ 1/sqrt(2N) スケール
        std = 0.02 / math.sqrt(2 * config.num_layers)
        for pn, p in self.named_parameters():
            if pn.endswith("proj.weight") or pn.endswith("w_down.weight"):
                nn.init.normal_(p, mean=0.0, std=std)

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:          # 全層bias=False なのでスキップ
                nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.padding_idx is not None:   #  padding_idxは未指定なのでスキップ
                with torch.no_grad():
                    module.weight[module.padding_idx].zero_()

    # static KV cache化対応した部分
    def forward(self, x, kv_cache=None, cache_position=None):
        bs, seq_len = x.shape
        # (1) 学習時
        if kv_cache is None:
            assert seq_len <= self.config.seq_len, "位置埋め込みの範囲外 seq_lenを超えています"
            positions = torch.arange(seq_len, device=x.device)
            attn_mask = None   # 学習時はmaskなし is_causal=True部分で作成
        # (2) 推論時
        else:
            # 【重要部分】キャッシュ利用時のマスク作成
            positions = cache_position          # cache_position：現在位置
            key_positions = torch.arange(kv_cache.max_cache_len, device=x.device)
            attn_mask = key_positions[None, :] <= cache_position[:, None]
            attn_mask = attn_mask[None, None, :, :]        # 形状：(1, 1, T, max_cache_len) or reshape(1,1,*attn_mask)          

        # 位置埋め込み
        x = self.token_embedding(x) + self.pos_embedding(positions).unsqueeze(0)
        x = self.dropout(x)

        # キャッシュは上書き更新される
        for block in self.transformer_blocks:
            x = block(x, kv_cache=kv_cache, cache_position=cache_position, attn_mask=attn_mask)

        # 推論時は最後のtokenのlogitsだけ使えばよい（decode時は seq_len=1 なので何もしない）
        if kv_cache is not None:
            x = x[:, -1:]           #　x[:, -1].unsqueeze(1)と同じ、わかりやす方でOK

        logits = self.fc(self.norm(x))
        return logits
```

```python
config = ModelConfig(tokenizer)
model = DNN(config).to(device)
```

### Random化

```python
class RandomGPTDataset(Dataset):
    def __init__(
        self,
        data_dir,
        tokenizer,
        context_size=128,
        samples_per_epoch=10_000,
    ):
        self.context_size = context_size
        self.samples_per_epoch = samples_per_epoch
        eod_id = tokenizer.token_to_id("<eod>")
        filenames = sorted(Path(data_dir).glob("*.txt"))

        all_ids = []
        for filename in filenames:
            text = filename.read_text(encoding="utf-8")
            document_ids = tokenizer.encode(text, add_special_tokens=False).ids
            all_ids.extend(document_ids)
            all_ids.append(eod_id)

        self.ids = torch.tensor(all_ids, dtype=torch.long)

    def __len__(self):
        # 1 epochあたりに何個のランダム窓を学習するか
        # 10,000個（回？）がデフォルト値
        return self.samples_per_epoch

    def __getitem__(self, idx):
        # idxは使わず、毎回ランダムな開始位置を選ぶ
        start = torch.randint(
            low=0,
            high=len(self.ids) - self.context_size,
            size=(1,),
        ).item()

        x = self.ids[start:start + self.context_size]
        y = self.ids[start + 1:start + self.context_size + 1]

        return {"ids": x, "labels": y}
```

```python
dataset = RandomGPTDataset(
    data_dir=data_dir,
    tokenizer=tokenizer,
    context_size=config.context_size,
    samples_per_epoch=10_000,
)

dataloader = DataLoader(
    dataset=dataset,
    batch_size=config.batch_size,  # メモリ足りない場合は小さくする
    shuffle=False,
    drop_last=True,
    num_workers=0, # 2,4のほうが早い？要検討
    pin_memory=True,
)
```

```python
decay, no_decay = [], []
for name, param in model.named_parameters():
    if not param.requires_grad:
        continue
    if param.dim() >= 2:
        decay.append(param)      # Linear の重み、Embedding
    else:
        no_decay.append(param)   # RMSNorm の gain、bias

optim_groups = [
    {"params": decay,    "weight_decay": config.weight_decay},
    {"params": no_decay, "weight_decay": 0.0},
]
optimizer = torch.optim.AdamW(optim_groups, lr=config.learning_rate, betas=(0.9, 0.95), fused=(device.type == "cuda"))
```

```python
criterion = nn.CrossEntropyLoss(ignore_index=config.ignore_index)    # 今回ignore_indexを利用していないがSFT用につけておく
```

```python
# step数で管理するためのデータローダー関数
def infinite_loader(dataloader):
    while True:
        for batch in dataloader:
            yield batch

data_iter = infinite_loader(dataloader)  # epochではなく、step数で計測
```

```python
from tqdm import tqdm
max_iters = config.num_steps  # step数
total_token = 0               # 累積トークン数を初期化
pbar = tqdm(range(max_iters))

use_bf16 = torch.cuda.is_available() and torch.cuda.is_bf16_supported()


model.train()                 # trainモードを明示
for step in pbar:
    batch = next(data_iter)
    input_ids = batch["ids"].to(device, non_blocking=True)
    labels = batch["labels"].to(device, non_blocking=True)
    optimizer.zero_grad()
    # (1) GPUが対応している場合は、bf16へ変更してflash attentionを使う
    with torch.autocast(device_type=device.type, dtype=torch.bfloat16, enabled=use_bf16):
        outputs = model(input_ids)
        loss = criterion(
            outputs.view(-1, config.vocab_size),
            labels.view(-1),
        )
    loss.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=config.max_grad_norm)  # 勾配クリップ
    optimizer.step()

    # このstepのトークン数 （バッチサイズ×系列長）を加算
    total_token += input_ids.shape[0] * input_ids.shape[1]

    # set_postfixは辞書が引数になる
    pbar.set_postfix({"loss": f"{loss.item():.4f}", "tokens": f"{total_token/1e6:.2f}M"})
    if (step+1) % 500 == 0:
        tqdm.write(f"{step+1}-step:\tloss:{loss.item():.3f}\ttokens:{total_token:,}")
```

### 文章生成

```python
@torch.inference_mode()
def generate_text(model, input_ids, max_new_tokens=100, eos_token_id=None,
                  kv_cache=None):
    model.eval()

    input_ids = input_ids.to(device)
    bs, prompt_len = input_ids.shape
    max_len = model.config.seq_len

    # 長さのチェック
    assert prompt_len < max_len, "プロンプトが seq_len 以上です"
    max_new_tokens = min(max_new_tokens, max_len - prompt_len)   # 合計長を seq_len 以下に

    # autocast の状態に応じて dtype を決定
    device_type = input_ids.device.type

    if torch.is_autocast_enabled(device_type):
        cache_dtype = torch.get_autocast_dtype(device_type)
    else:
        cache_dtype = next(model.parameters()).dtype

    # (1) kv cacheがない場合は領域確保
    if kv_cache is None:
        kv_cache = StaticKVCache(
            model.config,
            bs,
            device=input_ids.device,
            dtype=cache_dtype
        )
    # (2) すでにキャッシュがある場合は初期化    
    else:
        assert kv_cache.key_cache[0].dtype == cache_dtype
        kv_cache.reset()

    # (3) Prefill
    cache_position = torch.arange(prompt_len, device=input_ids.device)
    logits = model(input_ids, kv_cache=kv_cache, cache_position=cache_position)
    next_token = logits[:, -1, :].argmax(dim=-1, keepdim=True)
    generated = torch.cat([input_ids, next_token], dim=1)

    # (4) Decode
    cache_position = torch.tensor([prompt_len], device=input_ids.device)
    for _ in range(max_new_tokens - 1):
        if eos_token_id is not None and (next_token == eos_token_id).all():
            break
        # 最新の1トークンだけ入力
        logits = model(next_token, kv_cache=kv_cache, cache_position=cache_position)
        cache_position += 1
        next_token = logits[:, -1, :].argmax(dim=-1, keepdim=True)
        generated = torch.cat([generated, next_token], dim=1)
    return generated
```

```python
from tokenizers import decoders
import pandas as pd
tokenizer.decoder = decoders.ByteLevel() # BPEのときはこれが必要、文字化けするぞ

def response(prompt):
    model.eval()
    model_input = tokenizer.encode(prompt, add_special_tokens=False)
    input_ids = torch.tensor([model_input.ids], dtype=torch.long, device=device)

    output_ids = generate_text(
        model=model,
        input_ids=input_ids,
        max_new_tokens=config.seq_len,
        eos_token_id=config.eod_token_id,
    )

    # 全文をデコード
    generated_text = tokenizer.decode(output_ids[0].tolist(), skip_special_tokens=False)
    print(generated_text)

df = pd.read_csv("./data/sample_prompt.csv")

import time
start = time.perf_counter()
for num, prompt in enumerate(df["prompt"], 1):
    print(f"【{num}番目】")
    response(prompt)
end = time.perf_counter()
print(f"ループ全体の実行時間: {end - start :.3f} 秒")
```
